# Book-Crossing Dataset EDA

This notebook demonstrates exploratory data analysis on the Book-Crossing dataset using pandas.

```python
import pandas as pd
import numpy as np
import os

raw_dir = '../data/raw'
books = pd.read_csv(os.path.join(raw_dir, 'BX-Books.csv'), sep=';', encoding='latin-1', on_bad_lines='skip', low_memory=False)
ratings = pd.read_csv(os.path.join(raw_dir, 'BX-Book-Ratings.csv'), sep=';', encoding='latin-1')
users = pd.read_csv(os.path.join(raw_dir, 'BX-Users.csv'), sep=';', encoding='latin-1', on_bad_lines='skip')

print(f"Books: {len(books)}")
print(f"Ratings: {len(ratings)}")
print(f"Users: {len(users)}")
```

## Rating Distribution

```python
ratings['Book-Rating'].value_counts().sort_index().plot(kind='bar', title='Distribution of Ratings (0=Implicit)')
```
